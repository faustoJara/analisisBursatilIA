```python
import torch
import yfinance as yf
import pandas as pd
import numpy as np
from datetime import datetime


def script_unico_fase1(ticker="NVDA"):
    # 1. INGESTA DE DATOS (Múltiples fuentes)
    print(f"📥 Descargando datos históricos de {ticker} desde 1900...")
    df = yf.download(ticker, start="1900-01-01", progress=False)
    
    # 2. PROCESAMIENTO (Simulación de AWS Glue / NiFi)
    print("⚙️ Procesando variables técnicas...")
    
    # Ejemplo de variable independiente (Feature): Media móvil
    df['SMA_50'] = df['Close'].rolling(window=50).mean()
    
    # 3. DEFINICIÓN DE VARIABLE OBJETIVO (Target)
    # Calculamos el retorno porcentual para el día siguiente
    # Matematicamente: $$Return = \frac{Price_{t+1} - Price_{t}}{Price_{t}}$$
    df['Target_Return'] = df['Close'].pct_change().shift(-1)
    
    # 4. SIMULACIÓN DE INTEGRACIÓN CON MONGODB (NoSQL)
    # Aquí es donde entraría tu flujo de NiFi enviando sentimiento
    print("📝 Integrando scores de sentimiento desde MongoDB Atlas...")
    df['Sentiment_Score'] = np.random.uniform(-1, 1, size=len(df)) # Simulación inicial
    
    # Limpieza final
    df.dropna(inplace=True)
    
    print(f"✅ Repositorio Único creado. Filas totales: {len(df)}")
    return df

# Ejecutar
dataset_maestro = script_unico_fase1("NVDA")
print(dataset_maestro.tail())
```
