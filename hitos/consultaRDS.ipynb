```python
import pandas as pd
from sqlalchemy import create_engine
import os
from dotenv import load_dotenv

# Conexión a la nube
load_dotenv()
conexion_str = f"mysql+mysqlconnector://{os.getenv('DB_USER')}:{os.getenv('DB_PASSWORD')}@{os.getenv('DB_HOST')}:3306/{os.getenv('DB_NAME')}"
engine = create_engine(conexion_str)

print("=== AUDITORÍA DE LA BASE DE DATOS EN AWS ===\n")

# PRUEBA 1: Los días con mayor caos informativo
print("📊 1. TOP 5 DÍAS CON MÁS NOTICIAS EN LA HISTORIA DE NVIDIA:")
query_top = "SELECT fecha, close_price, sentiment_score, news_volume FROM dataset_nvda_unificado ORDER BY news_volume DESC LIMIT 5"
print(pd.read_sql(query_top, con=engine))

print("\n--------------------------------------------------\n")

# PRUEBA 2: Los días de silencio absoluto (Comprobando la inserción de ceros)
print("🤫 2. MUESTRA DE 5 DÍAS SIN NINGUNA NOTICIA (Debería haber muchos en los años 2000):")
query_ceros = "SELECT fecha, close_price, sentiment_score, news_volume FROM dataset_nvda_unificado WHERE news_volume = 0 LIMIT 5"
print(pd.read_sql(query_ceros, con=engine))
```
